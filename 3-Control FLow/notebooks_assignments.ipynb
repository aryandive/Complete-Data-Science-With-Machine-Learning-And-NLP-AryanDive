### Assignment 3: if-elif-else Statement

Write a program that asks the user to input a number and prints whether the number is positive, negative, or zero.

```python
a = int(input("Enter a number to check it is positive or negative: "))

if a > 0:
    print(f"The number {a} is a positive number")
elif a == 0:
    print(f"The number {a} is zero")
else:
    print(f"The number {a} is a negative number")
            
```

### Assignment 4: Nested if Statement

Write a program that asks the user to input a number and prints whether the number is positive and even, positive and odd, or negative.

```python
a = int(input("Enter a number to check it is positive or negative: "))

if a > 0:
    if a % 2 == 0:
        print(f"The number {a} is a positive even number")
    else:
        print(f"The number {a} is a positive odd number")
elif a == 0:
    print(f"The number {a} is zero")
else:
    print(f"The number {a} is a negative number")
            
```

### Assignment 8: break Statement

Write a program that asks the user to input numbers until they input 0. The program should print the sum of all the input numbers.

```python
total = 0

while True:
    number = float(input("Enter a number a number between 0 to any"))
    if number == 0:
        break
    total += number
print(f"The sum of all the numbers is {total}.")
```

### Assignment 12: Factorial Calculation

Write a program that calculates the factorial of a number input by the user using a while loop.

```python
number = int(input("Enter a number: "))
factorial = 1
i = 1
while i <= number:
    factorial *= i
    i += 1
print(f"The factorial of {number} is {factorial}.")
```

### Assignment 14: Prime Number Check

Write a program that checks if a number input by the user is a prime number using a for loop.

```python
number = int(input("Enter your number: "))

is_Prime = True

if number <= 1:
    is_Prime = False
else: 
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            is_Prime = False
        break

if is_Prime:
    print(f"The number {number} is a prime number")
else:
    print(f"The number {number} is not a prime number")
```

### Assignment 15: Fibonacci Sequence

Write a program that prints the first n Fibonacci numbers, where n is input by the user.

```python
n = int(input("Enter the number of Fibonacci numbers to print: "))
a, b = 0, 1
count = 0
while count < n:
    print(a)
    a, b = b, a + b
    count += 1
```
