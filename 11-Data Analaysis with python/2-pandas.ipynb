Pandas is a fast, powerful, and open-source Python library used for data manipulation and analysis. It provides highly flexible data structures that allow you to work easily with structured, tabular data—much like a digital spreadsheet or an SQL database table. The name originates from the econometrics term "Panel Data".

Pandas relies primarily on two data structures to handle data:
Series: A one-dimensional labeled array that can hold any data type (e.g., a single column of numbers or strings).
DataFrame: A two-dimensional, table-like data structure with labeled rows and columns (the most frequently used Pandas object).

```python
import pandas as pd
```

```python
## Series
## A Pandas Series is a one-dimensional labeled array capable of holding any data type (integers, strings, floating point numbers, Python objects, etc.). 
## It is similar to a column in a spreadsheet or a SQL table. Each element in a Series has an associated label, known as an index.

import pandas as pd
data = [1, 2, 3, 4, 5]
series =pd.Series(data)
print ("Series: \n", series)
print(type(series))
```

```python
## Create a series from dictionary
data={'a': 1, 'b': 2, 'c': 3}
series = pd.Series(data)
print(series)
```

```python
data=[10, 20, 30]
index=['a', 'b', 'c']
pd.Series(data, index=index)
```

```python
## DataFrame
## Create a DataFrame from a dictionary of lists
data = {
    'Name': ['Aryan', 'Bob', 'Charlie'],
    'Age': [21, 30, 35],
    'City': ['Pune', 'Los Angeles', 'Chicago']
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
## Create a Data Frame from a list of dictionaries
import pandas as pd

data = [
    {'Name': 'Aryan', 'Age': 21, 'City': 'Pune'},
    {'Name': 'Bob', 'Age': 30, 'City': 'Los Angeles'},
    {'Name': 'Charlie', 'Age': 35, 'City': 'Chicago'},
    {'Name': 'David', 'Age': 28, 'City': 'New York'}
]

df = pd.DataFrame(data)
print(df)
print(type(df))  
```
